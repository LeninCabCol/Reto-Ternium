```python
# ============================================================
# BLOQUE 1: Distribución de tarimas válidas por NUMERO_PARTE_HIJO
# Queremos saber si hay concentración (pocos números de parte
# con muchas tarimas) o distribución uniforme
# ============================================================

# Filtramos: tarimas no desarmadas y mismo MATERIAL_PADRE dentro de cada tarima
tarimas_validas = df[df['DESARMA_TARIMA'].isna()].copy()

# Identificar tarimas que tienen más de un MATERIAL_PADRE (las excepciones a eliminar)
tarimas_multi_padre = (
    tarimas_validas.groupby('ARMA_TARIMA')['MATERIAL_PADRE']
    .nunique()
)
tarimas_a_eliminar = tarimas_multi_padre[tarimas_multi_padre > 1].index

# Quedarnos solo con tarimas "limpias"
tarimas_validas = tarimas_validas[~tarimas_validas['ARMA_TARIMA'].isin(tarimas_a_eliminar)]

print(f"Tarimas válidas (no desarmadas, mismo MATERIAL_PADRE): {tarimas_validas['ARMA_TARIMA'].nunique()}")
print(f"NUMERO_PARTE_HIJO distintos en tarimas válidas: {tarimas_validas['NUMERO_PARTE_HIJO'].nunique()}")

# Contar cuántas tarimas únicas tiene cada número de parte
tarimas_por_parte = (
    tarimas_validas.groupby('NUMERO_PARTE_HIJO')['ARMA_TARIMA']
    .nunique()
    .sort_values(ascending=False)
    .reset_index()
)
tarimas_por_parte.columns = ['NUMERO_PARTE_HIJO', 'n_tarimas']

# Distribución por rangos
bins = [0, 1, 3, 5, 10, 20, 50, float('inf')]
labels = ['1', '2-3', '4-5', '6-10', '11-20', '21-50', '50+']
tarimas_por_parte['rango'] = pd.cut(tarimas_por_parte['n_tarimas'], bins=bins, labels=labels)

resumen = tarimas_por_parte.groupby('rango', observed=True)['NUMERO_PARTE_HIJO'].count()
resumen_pct = (resumen / resumen.sum() * 100).round(1)

print("\nDistribución de números de parte por cantidad de tarimas válidas:")
print(pd.DataFrame({'n_numeros_parte': resumen, 'porcentaje': resumen_pct}))
```

```python
# ============================================================
# BLOQUE 2: ¿Cuántos NUMERO_PARTE_HIJO ya tienen valores conocidos
# en PIEZAS_POR_PAQ_HIJO y PESO_MAX_DESPACHO_HIJO?
# Solo necesitamos estimar los que tienen NOA
# ============================================================

# Reemplazar 'NOA' por NaN para tratarlos como faltantes
df_check = df.copy()
df_check['PIEZAS_POR_PAQ_HIJO'] = df_check['PIEZAS_POR_PAQ_HIJO'].replace('NOA', np.nan)
df_check['PESO_MAX_DESPACHO_HIJO'] = df_check['PESO_MAX_DESPACHO_HIJO'].replace('NOA', np.nan)

# Para cada NUMERO_PARTE_HIJO, ver si tiene al menos un valor conocido
resumen_partes = df_check.groupby('NUMERO_PARTE_HIJO').agg(
    tiene_piezas=('PIEZAS_POR_PAQ_HIJO', lambda x: x.notna().any()),
    tiene_peso_max=('PESO_MAX_DESPACHO_HIJO', lambda x: x.notna().any()),
    total_cintas=('ARMA_TARIMA', 'count')
).reset_index()

print("=== PIEZAS_POR_PAQ_HIJO ===")
print(resumen_partes['tiene_piezas'].value_counts().rename({True: 'Tiene valor conocido', False: 'Solo NOA'}))

print("\n=== PESO_MAX_DESPACHO_HIJO ===")
print(resumen_partes['tiene_peso_max'].value_counts().rename({True: 'Tiene valor conocido', False: 'Solo NOA'}))

print("\n=== Combinación de ambas ===")
resumen_partes['categoria'] = 'Solo NOA en ambas'
resumen_partes.loc[resumen_partes['tiene_piezas'] & resumen_partes['tiene_peso_max'], 'categoria'] = 'Ambas conocidas'
resumen_partes.loc[resumen_partes['tiene_piezas'] & ~resumen_partes['tiene_peso_max'], 'categoria'] = 'Solo piezas conocida'
resumen_partes.loc[~resumen_partes['tiene_piezas'] & resumen_partes['tiene_peso_max'], 'categoria'] = 'Solo peso conocido'

print(resumen_partes['categoria'].value_counts())
```

```python
# ============================================================
# BLOQUE 3: Para los que SÍ tienen valor conocido,
# ¿es consistente ese valor o varía dentro del mismo número de parte?
# Esto nos dice qué tan confiables son los valores existentes
# ============================================================

# Solo filas con valor conocido en PIEZAS_POR_PAQ_HIJO
piezas_conocidas = df_check[df_check['PIEZAS_POR_PAQ_HIJO'].notna()]

consistencia_piezas = piezas_conocidas.groupby('NUMERO_PARTE_HIJO')['PIEZAS_POR_PAQ_HIJO'].nunique()
print("=== Consistencia de PIEZAS_POR_PAQ_HIJO ===")
print("Valores distintos por número de parte:")
print(consistencia_piezas.value_counts().rename_axis('n_valores_distintos').reset_index(name='n_numeros_parte'))

# Solo filas con valor conocido en PESO_MAX_DESPACHO_HIJO
peso_conocido = df_check[df_check['PESO_MAX_DESPACHO_HIJO'].notna()]

consistencia_peso = peso_conocido.groupby('NUMERO_PARTE_HIJO')['PESO_MAX_DESPACHO_HIJO'].nunique()
print("\n=== Consistencia de PESO_MAX_DESPACHO_HIJO ===")
print("Valores distintos por número de parte:")
print(consistencia_peso.value_counts().rename_axis('n_valores_distintos').reset_index(name='n_numeros_parte'))
```

```python
# ============================================================
# BLOQUE 4: Homogeneidad interna de NUMERO_PARTE_HIJO
# Queremos saber si las cintas del mismo número de parte
# tienen dimensiones consistentes o muy variables
# ============================================================

# Asegurarnos de que df_check existe (puede venir del BLOQUE 2 o se recrea aquí)
if 'df_check' not in dir():
    df_check = df.copy()
    df_check['PIEZAS_POR_PAQ_HIJO'] = df_check['PIEZAS_POR_PAQ_HIJO'].replace('NOA', np.nan)
    df_check['PESO_MAX_DESPACHO_HIJO'] = df_check['PESO_MAX_DESPACHO_HIJO'].replace('NOA', np.nan)

# Asegurarnos de trabajar con valores numéricos
df_check['ANCHO_HIJO'] = pd.to_numeric(df_check['ANCHO_HIJO'], errors='coerce')
df_check['ESPESOR_HIJO'] = pd.to_numeric(df_check['ESPESOR_HIJO'], errors='coerce')

# Para cada número de parte, calcular rango (max - min) de cada dimensión
homogeneidad = df_check.groupby('NUMERO_PARTE_HIJO').agg(
    n_cintas=('ANCHO_HIJO', 'count'),
    ancho_min=('ANCHO_HIJO', 'min'),
    ancho_max=('ANCHO_HIJO', 'max'),
    ancho_rango=('ANCHO_HIJO', lambda x: x.max() - x.min()),
    ancho_valores_distintos=('ANCHO_HIJO', 'nunique'),
    espesor_min=('ESPESOR_HIJO', 'min'),
    espesor_max=('ESPESOR_HIJO', 'max'),
    espesor_rango=('ESPESOR_HIJO', lambda x: x.max() - x.min()),
    espesor_valores_distintos=('ESPESOR_HIJO', 'nunique'),
).reset_index()

# ¿Cuántos números de parte tienen exactamente 1 valor de ancho y 1 de espesor?
perfectamente_homogeneos = homogeneidad[
    (homogeneidad['ancho_valores_distintos'] == 1) & 
    (homogeneidad['espesor_valores_distintos'] == 1)
]
print(f"Números de parte con ancho Y espesor completamente fijos: "
      f"{len(perfectamente_homogeneos)} de {len(homogeneidad)} "
      f"({len(perfectamente_homogeneos)/len(homogeneidad)*100:.1f}%)")

# Distribución del rango de ancho
print("\n=== Variación de ANCHO_HIJO por número de parte (mm) ===")
bins_ancho = [-1, 0, 1, 5, 10, float('inf')]
labels_ancho = ['Fijo (0)', 'Hasta 1mm', '1-5mm', '5-10mm', 'Más de 10mm']
homogeneidad['rango_ancho_cat'] = pd.cut(homogeneidad['ancho_rango'], bins=bins_ancho, labels=labels_ancho)
print(homogeneidad['rango_ancho_cat'].value_counts().sort_index())

# Distribución del rango de espesor
print("\n=== Variación de ESPESOR_HIJO por número de parte (mm) ===")
bins_esp = [-1, 0, 0.1, 0.5, 1, float('inf')]
labels_esp = ['Fijo (0)', 'Hasta 0.1mm', '0.1-0.5mm', '0.5-1mm', 'Más de 1mm']
homogeneidad['rango_espesor_cat'] = pd.cut(homogeneidad['espesor_rango'], bins=bins_esp, labels=labels_esp)
print(homogeneidad['rango_espesor_cat'].value_counts().sort_index())

# Ver los números de parte más heterogéneos (los problemáticos)
print("\n=== Top 10 números de parte con mayor variación en ANCHO_HIJO ===")
print(homogeneidad.nlargest(10, 'ancho_rango')[
    ['NUMERO_PARTE_HIJO', 'n_cintas', 'ancho_min', 'ancho_max', 'ancho_rango', 'espesor_rango']
].to_string(index=False))
```

```python
# ============================================================
# BLOQUE 5: ¿Qué tan relevantes son los números de parte
# con ancho variable (>1mm)? ¿Cuántas cintas representan?
# ============================================================

# Identificar números de parte problemáticos (variación de ancho > 1mm)
partes_problematicas = homogeneidad[homogeneidad['ancho_rango'] > 1]['NUMERO_PARTE_HIJO']
partes_ok = homogeneidad[homogeneidad['ancho_rango'] <= 1]['NUMERO_PARTE_HIJO']

total_cintas = len(df_check)
cintas_problematicas = df_check[df_check['NUMERO_PARTE_HIJO'].isin(partes_problematicas)]
cintas_ok = df_check[df_check['NUMERO_PARTE_HIJO'].isin(partes_ok)]

print("=== Impacto de números de parte con ancho variable (>1mm) ===")
print(f"Números de parte problemáticos: {len(partes_problematicas)} de {len(homogeneidad)} ({len(partes_problematicas)/len(homogeneidad)*100:.1f}%)")
print(f"Cintas en partes problemáticas: {len(cintas_problematicas):,} de {total_cintas:,} ({len(cintas_problematicas)/total_cintas*100:.1f}%)")
print(f"Cintas en partes OK:            {len(cintas_ok):,} de {total_cintas:,} ({len(cintas_ok)/total_cintas*100:.1f}%)")

# De las cintas problemáticas, ¿cuántas están en tarimas desarmadas vs no desarmadas?
print("\n=== Dentro de las cintas problemáticas ===")
print(cintas_problematicas['DESARMA_TARIMA'].isna().value_counts()
      .rename({True: 'No desarmadas', False: 'Desarmadas'}))

# ¿Las partes problemáticas tienen más desarmes que las OK?
tasa_desarme_problematicas = cintas_problematicas['DESARMA_TARIMA'].notna().mean()
tasa_desarme_ok = cintas_ok['DESARMA_TARIMA'].notna().mean()

print(f"\nTasa de desarme en partes problemáticas: {tasa_desarme_problematicas*100:.1f}%")
print(f"Tasa de desarme en partes OK:            {tasa_desarme_ok*100:.1f}%")
```

se va a excluir las cintas que vienen del mismo numero parte hijo y tienen dimensiones muy diferentes
