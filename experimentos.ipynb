En el dataframe "dF" dejamos unicamente las variables que nos sirven para identificar los patrones entre las variables que se quieren probar


```python
variables_utiles = ['ID_ODE', 'CONSIGNATARIO','SOLICITANTE','C_ITEM','PESO_BRUTO_DESP','C_PEDIDO',
                    'MATERIAL_PADRE',
                    'PESO_MATERIAL_SALIDA','DESCRIPCION_EMBALAJE',
                    'NUMERO_PARTE_HIJO','PIEZAS_POR_PAQ_HIJO',
                    'PESO_MAX_DESPACHO_HIJO','ARMA_TARIMA',
                    'DESARMA_TARIMA']



df = df[variables_utiles]
```

Se quiere probar que todas las cintas que pertenecen a una misma orden pertenecen a un mismo C_PEDIDO

```python
# Para cada ID_ODE, verificar si todos sus valores de C_PEDIDO son iguales
grupos = df.groupby('ID_ODE')['C_PEDIDO'].nunique()

# Grupos donde ID_ODE tiene más de un valor distinto de C_PEDIDO
inconsistentes = grupos[grupos > 1]

if inconsistentes.empty:
    print("CONFIRMADO: Cada ID_ODE siempre tiene el mismo valor de C_PEDIDO en todas sus filas.")
else:
    print(f"NO SE CUMPLE: {len(inconsistentes)} ID_ODE(s) tienen más de un valor distinto en C_PEDIDO:")
    print(inconsistentes)
```

Se crea un nuevo dataframe ordenado por ordenes para ver cómo es que se realiza una orden y como se termina por desarmar las tarimas

Se crea un datafreame con unicamente las cintas que se armo tarima y no se desarmo la tarima

```python
df_pq_desarme = df_desarme
variables_compaDesarme = [
                    'MATERIAL_PADRE',
                    'NUMERO_PARTE_HIJO',
                    'ARMA_TARIMA',
                    'DESARMA_TARIMA']



df_pq_desarme = df_pq_desarme[variables_compaDesarme]
df_pq_desarme = df_pq_desarme.sort_values('ARMA_TARIMA', ascending=False)
df_pq_desarme = df_pq_desarme[df_pq_desarme['DESARMA_TARIMA'].isna()]
df_pq_desarme
```

Volvemos a ver si existen aun tarimas que se armaron (pero no se desarmaron despues) que tengan diferente material padre, vemos que aun hay

```python
grupos = df_pq_desarme.groupby('ARMA_TARIMA')['MATERIAL_PADRE'].nunique()


inconsistentes = grupos[grupos > 1]

if inconsistentes.empty:
    print("CONFIRMADO: Cada ARMA_TARIMA siempre tiene el mismo valor de MATERIAL_PADRE en todas sus filas.")
else:
    print(f"NO SE CUMPLE: {len(inconsistentes)} ARMA_TARIMA(s) tienen más de un valor distinto en MATERIAL_PADRE:")
    print(inconsistentes)
```

Ahora vemos si existen tarimas que se armaron y no se desarmaron despues que tengan cintas que vengan de diferente producto. vemos que hay muchisimas menos (de 277 tarimas que se armaron con productos distintos la mayoria se desarmó y solo quedan 3)

Esto prueba que en una misma tarima no pueden ir cintas que vengan de difernte producto

```python
grupos = df_pq_desarme.groupby('ARMA_TARIMA')['NUMERO_PARTE_HIJO'].nunique()


inconsistentes = grupos[grupos > 1]

if inconsistentes.empty:
    print("CONFIRMADO: Cada ARMA_TARIMA siempre tiene el mismo valor de NUMERO_PARTE_HIJO en todas sus filas.")
else:
    print(f"NO SE CUMPLE: {len(inconsistentes)} ARMA_TARIMA(s) tienen más de un valor distinto en NUMERO_PARTE_HIJO:")
    print(inconsistentes)
```
